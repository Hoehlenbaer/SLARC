```python
def repvgg_model_convert(model: torch.nn.Module, save_path=None):
    """
    Faltet alle RepConv-Blöcke im Modell zusammen.
    Ruf dies VOR dem Export nach ONNX auf!
    """
    model.eval() # Sehr wichtig!
    
    # Durchsuche das ganze Modell nach RepConv-Modulen
    converted_count = 0
    for module in model.modules():
        if isinstance(module, RepConv):
            module.switch_to_deploy()
            converted_count += 1
            
    print(f"✅ {converted_count} RepConv-Blöcke erfolgreich gefaltet!")
    print(f"Das Modell ist jetzt zu 100% NPU-kompatibel und Inferenz-optimiert.")
    
    if save_path:
        torch.save(model.state_dict(), save_path)
    return model

```
